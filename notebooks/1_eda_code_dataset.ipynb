# CodeRoast 🔥 — Notebook 1: Exploratory Data Analysis (EDA)

This notebook performs exploratory analysis on the multilingual GitHub code dataset collected for **CodeRoast**.

### Objectives:
1. Inspect dataset structure and language breakdown
2. Analyze class distributions across quality levels (Pristine, Acceptable, Concerning, Disaster)
3. Compute static metrics (LOC, Cyclomatic Complexity, Comment Ratio)
4. Visualize key relationships between code features and quality ratings

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot styling
plt.style.use('dark_background')
sns.set_theme(style='darkgrid')

# Path configuration
DATA_PATH = '../data/processed/code_samples.csv'
if not os.path.exists(DATA_PATH):
    DATA_PATH = 'data/processed/code_samples.csv'

print(f"Loading dataset from: {DATA_PATH}")
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
# Dataset Summary
print("Dataset Shape:", df.shape)
print("\nLanguage Distribution:\n", df['language'].value_counts())
print("\nQuality Label Distribution:\n", df['quality_label'].value_counts())
```

```python
# Visualizing Quality Class Balance
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

quality_names = ['Pristine', 'Acceptable', 'Concerning', 'Disaster']
counts = df['quality_label'].value_counts().sort_index()

sns.barplot(x=[quality_names[i] for i in counts.index], y=counts.values, ax=ax[0], palette='flare')
ax[0].set_title('Code Quality Class Distribution', fontsize=14, fontweight='bold')
ax[0].set_ylabel('Sample Count')

lang_counts = df['language'].value_counts()
ax[1].pie(lang_counts.values, labels=lang_counts.index, autopct='%1.1f%%', colors=['#ff4b4b', '#ff8c00', '#ffd700'])
ax[1].set_title('Language Share', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Feature Extraction — Lines of Code & Code Length
df['char_length'] = df['code'].apply(len)
df['line_count'] = df['code'].apply(lambda x: len(str(x).splitlines()))

plt.figure(figsize=(10, 4))
sns.histplot(data=df, x='line_count', hue='language', kde=True, bins=30, palette='magma')
plt.title('Lines of Code (LOC) Distribution per Language', fontsize=14, fontweight='bold')
plt.xlabel('Lines of Code')
plt.show()
```

### Key EDA Takeaways:
- The dataset contains clean, pre-processed samples across **Python, Java, and JavaScript**.
- Class distribution is balanced to prevent model bias towards pristine or disaster code.
- Feature lengths show distinct structural patterns suitable for n-gram TF-IDF and sequence modeling.
