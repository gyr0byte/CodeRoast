```python
import os
import sys
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, '..')
from src.ml.codebert_model import CodeBERTScorer
from src.ml.lstm_model import CodeTokenizer, LSTMSeverityModel

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using PyTorch Device: {device}")
```

```python
# Load dataset
df = pd.read_csv('../data/processed/code_samples.csv')
print(f"Dataset samples: {len(df)}")
df[['code', 'language', 'quality_label', 'severity_score']].head()
```

```python
# 1. Evaluate Hugging Face CodeBERT
print("Initializing Hugging Face CodeBERT (microsoft/codebert-base)...------------- ")
bert_scorer = CodeBERTScorer()

test_code = "def process(x):\n    for i in range(len(x)):\n        for j in range(len(x)):\n            pass"
quality, severity = bert_scorer.predict_quality_and_severity(test_code)
print(f"Test Code Prediction -> Quality Level: {quality}, Severity Score: {severity:.4f}")
```

```python
# 2. PyTorch LSTM Tokenization & Model Setup
tokenizer = CodeTokenizer(max_vocab_size=5000, max_length=200)
X = tokenizer.fit_transform(df['code'].tolist())
y = df['severity_score'].values.astype(np.float32)

lstm = LSTMSeverityModel(vocab_size=tokenizer.vocab_size, max_length=tokenizer.max_length)
print("PyTorch LSTM Model Architecture:\n", lstm.model)
```
